```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load dataset
df = pd.read_csv('social_media_cleaned_dataset_10k.csv')

# 2. Check first 5 rows
print(df.head())

# 3. Platform-wise Engagement Rate Bar Plot
plt.figure(figsize=(8, 5))
sns.barplot(x='platform', y='engagement_rate', data=df, palette='viridis')
plt.title('Average Engagement Rate by Platform')
plt.show()
```

```python
# 4. Correlation Heatmap (Numeric columns-oda relationship)
plt.figure(figsize=(10, 6))
sns.heatmap(df[['follower_count', 'likes', 'comments', 'shares', 'reach', 'engagement_rate']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Social Media Metrics')
plt.show()
```
